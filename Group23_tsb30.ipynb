## Problem 1: String alphabet analysis

```python
import string
alphabet = string.ascii_lowercase
print(alphabet)
```

```python
# (1) A string of English text, exactly 300 characters long, containing some numbers

text = ('In 1969, the Apollo 11 mission carried 3 astronauts to the Moon. About 600 million '
        'people watched the landing on television. The crew spent 21 hours on the surface and '
        'brought back 47 pounds of rocks. Today, more than 50 years later, scientists still study '
        'those samples in labs all around the world.')

print('The length of the string is {} characters.'.format(len(text)))
words = text.split()
print('The number of words in the string is {}.'.format(len(words)))
```

```python
# (2) Reverse the string, replace 'b' with 'X', and make everything lower case.
# Lower case is applied first, so that any capital 'B' is also replaced.
# 'X' is added after lower(), so it stays upper case as asked.

reversed_text = text[::-1].lower().replace('b', 'X')
print('\nReversed string, lower case, with b replaced by X:\n')
print(reversed_text)
```

```python
# (3) Count the letter 'a', using the format method

lower_text = text.lower()
print('\nThe letter {} appears {} times in the string.'.format('a', lower_text.count('a')))

# Do the same for the remaining letters with a for loop, and collect missing letters
missing = []
for letter in alphabet[1:]:
    n = lower_text.count(letter)
    print('The letter {} appears {} times in the string.'.format(letter, n))
    if n == 0:
        missing.append(letter)

# also check 'a' so the missing list covers the whole alphabet
if lower_text.count('a') == 0:
    missing.insert(0, 'a')
print('\nLetters missing from the string: {}'.format(missing))
```

```python
# (4) Remove digits 0 to 8, and replace 9 with the word 'nine'

new_text = text
for digit in '012345678':
    new_text = new_text.replace(digit, '')
new_text = new_text.replace('9', 'nine')
print('\nString with digits 0-8 removed and 9 replaced by nine:\n')
print(new_text)
```

## Problem 2. Dictionary and breaking the loop

```python
import random

# A string with 10 different words

s = 'the quick brown fox jumps over a lazy sleeping dog'
```

```python
# (1) Dictionary: key = position of the word (starting at 0), value = the word

words = s.split()
d = {}
for i in range(len(words)):
    d[i] = words[i]
print('d =', d)
```

```python
# (2) Draw 20 random keys from 0 to 9 and build a new sentence from them

s_new = ''
for i in range(20):
    key = random.randint(0, 9)
    s_new = s_new + d[key] + ' '
s_new = s_new.strip()
print('\ns_new = {}'.format(s_new))
```

```python
# (3) Loop over d and count how many times each word appears in s_new.
# Split s_new into a list, so whole words are counted
# (e.g. 'a' is not counted inside 'lazy').

new_words = s_new.split()
for key in d:
    word = d[key]
    n = new_words.count(word)
    if n == 0:
        print("'{}' is missing in s_new".format(word))
        break
    print("The word '{}' appears {} times in s_new.\n".format(word, n))
```
